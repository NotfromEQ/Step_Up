```python
import matplotlib.pyplot as plt
from scipy.io import loadmat
```

```python
import kagglehub

# Download latest version
digit_dataset = kagglehub.dataset_download("avnishnish/mnist-original")

print("Path to dataset files:", digit_dataset)
```

```python
import os

mnist_raw = loadmat(os.path.join(digit_dataset, "mnist-original.mat"))
mnist={
    "data" : mnist_raw["data"].T,
    "target" : mnist_raw["label"][0]
}
```

```python
x,y = mnist["data"],mnist["target"]
number = x[15000]
number_image = number.reshape(28,28)
```

```python
plt.imshow(number_image,cmap=plt.cm.berlin,interpolation="nearest")
plt.show
print(y[15000])
```
