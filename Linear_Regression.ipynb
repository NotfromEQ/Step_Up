```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(-5,5,100)
y = 2*x+1
```

```python
plt.plot(x,y,'r',label='y=2x+1')
plt.xlabel('x')
plt.ylabel('y')
plt.grid()
plt.show()
```
